```python
# Import library yang diperlukan
import numpy as np
import matplotlib.pyplot as plt

# Fungsi untuk menggambar jajargenjang
def gambar_jajargenjang(titik, ax, warna='b', label='Jajargenjang'):
    # Menggambar jajargenjang berdasarkan titik yang diberikan
    jajargenjang = plt.Polygon(titik, closed=True, fill=None, edgecolor=warna, linewidth=2)
    ax.add_patch(jajargenjang)
    ax.plot(titik[:, 0], titik[:, 1], 'o', color=warna, markersize=6)

# Fungsi untuk melakukan rotasi
def rotasi(titik, sudut):
    # Menghitung sudut dalam radian
    radian = np.radians(sudut)
    
    # Matriks rotasi
    rotasi_matrix = np.array([[np.cos(radian), -np.sin(radian)],
                              [np.sin(radian), np.cos(radian)]])
    
    # Melakukan rotasi pada setiap titik
    titik_rotated = np.dot(titik, rotasi_matrix)
    return titik_rotated

# Titik-titik jajargenjang (langsung diinput)
# Format: [x, y] untuk setiap titik
titik_asli = np.array([[1, 1], [4, 1], [5, 3], [2, 3]])

# Sudut rotasi 180 derajat
sudut_rotasi = 180

# --- Membuat plot dengan 2 subplot ---
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 8))

# --- Plot 1: Jajargenjang Sebelum Rotasi ---
ax1.set_xlim(-6, 6)
ax1.set_ylim(-6, 6)
ax1.axhline(0, color='black', linewidth=0.5, ls='--')
ax1.axvline(0, color='black', linewidth=0.5, ls='--')
ax1.grid(color='gray', linestyle='--', linewidth=0.5)

gambar_jajargenjang(titik_asli, ax1, 'blue', 'Jajargenjang Asli')
ax1.set_title('Jajargenjang Sebelum Rotasi', fontsize=14, fontweight='bold')
ax1.set_aspect('equal', adjustable='box')
ax1.set_xlabel('X')
ax1.set_ylabel('Y')

# --- Melakukan rotasi ---
titik_rotated = rotasi(titik_asli, sudut_rotasi)

# --- Plot 2: Jajargenjang Setelah Rotasi ---
ax2.set_xlim(-6, 6)
ax2.set_ylim(-6, 6)
ax2.axhline(0, color='black', linewidth=0.5, ls='--')
ax2.axvline(0, color='black', linewidth=0.5, ls='--')
ax2.grid(color='gray', linestyle='--', linewidth=0.5)

gambar_jajargenjang(titik_rotated, ax2, 'red', 'Jajargenjang Rotasi')
ax2.set_title(f'Jajargenjang Setelah Rotasi {sudut_rotasi}°', fontsize=14, fontweight='bold')
ax2.set_aspect('equal', adjustable='box')
ax2.set_xlabel('X')
ax2.set_ylabel('Y')

plt.tight_layout()
plt.show()

# Menampilkan koordinat titik
print("Koordinat Jajargenjang Asli:")
for i, titik in enumerate(titik_asli):
    print(f"  Titik {i+1}: ({titik[0]}, {titik[1]})")

print(f"\nKoordinat Jajargenjang Setelah Rotasi {sudut_rotasi}°:")
for i, titik in enumerate(titik_rotated):
    print(f"  Titik {i+1}: ({titik[0]:.2f}, {titik[1]:.2f})")
```
