```python
# Import library yang diperlukan
import numpy as np
import matplotlib.pyplot as plt

# Fungsi untuk menggambar segitiga
def gambar_segitiga(titik, ax):
    # Menggambar segitiga berdasarkan titik yang diberikan
    # closed=True agar garis menutup kembali ke titik awal
    segitiga = plt.Polygon(titik, closed=True, fill=None, edgecolor='b', linewidth=2)
    ax.add_patch(segitiga)

# Fungsi untuk melakukan rotasi
def rotasi(titik, sudut):
    # Menghitung sudut dalam radian
    radian = np.radians(sudut)

    # Matriks rotasi
    rotasi_matrix = np.array([[np.cos(radian), -np.sin(radian)],
                              [np.sin(radian), np.cos(radian)]])

    # Melakukan rotasi pada setiap titik segitiga
    titik_rotated = np.dot(titik, rotasi_matrix)
    return titik_rotated

# Titik-titik segitiga (langsung diinput)
titik_asli = np.array([[2, 1], [5, 1], [3.5, 4]])

# Sudut rotasi (dalam derajat)
sudut_rotasi = 45

# --- Plot 1: Segitiga Sebelum Rotasi ---
# Menggambar segitiga sebelum rotasi
fig, ax = plt.subplots(figsize=(8, 8))
ax.set_xlim(-10, 10)
ax.set_ylim(-10, 10)
ax.axhline(0, color='black', linewidth=0.5, ls='--')
ax.axvline(0, color='black', linewidth=0.5, ls='--')
ax.grid(color='gray', linestyle='--', linewidth=0.5)

# Gambar segitiga asli
gambar_segitiga(titik_asli, ax)
ax.set_title('Segitiga Sebelum Rotasi')
ax.set_aspect('equal', adjustable='box')
plt.show()

# --- Proses Rotasi ---
# Melakukan rotasi
titik_rotated = rotasi(titik_asli, sudut_rotasi)

# --- Plot 2: Segitiga Setelah Rotasi ---
# Menggambar segitiga setelah rotasi
fig, ax = plt.subplots(figsize=(8, 8))
ax.set_xlim(-10, 10)
ax.set_ylim(-10, 10)
ax.axhline(0, color='black', linewidth=0.5, ls='--')
ax.axvline(0, color='black', linewidth=0.5, ls='--')
ax.grid(color='gray', linestyle='--', linewidth=0.5)

# Gambar segitiga yang sudah dirotasi
gambar_segitiga(titik_rotated, ax)
ax.set_title(f'Segitiga Setelah Rotasi ({sudut_rotasi} Derajat)')
ax.set_aspect('equal', adjustable='box')
plt.show()
```

```python
# Import library yang diperlukan
import numpy as np
import matplotlib.pyplot as plt

# Fungsi untuk menggambar lingkaran
def gambar_lingkaran(radius, center, ax, label='Lingkaran'):
    # Membuat array sudut dari 0 hingga 2*pi
    theta = np.linspace(0, 2 * np.pi, 100)

    # Menghitung koordinat x dan y dari lingkaran
    x = radius * np.cos(theta) + center[0]
    y = radius * np.sin(theta) + center[1]

    # Menggambar lingkaran
    ax.plot(x, y, label=label, linewidth=2)

# Fungsi untuk melakukan rotasi pada pusat lingkaran
def rotasi_lingkaran(center, sudut):
    # Menghitung sudut dalam radian
    radian = np.radians(sudut)

    # Matriks rotasi
    rotasi_matrix = np.array([[np.cos(radian), -np.sin(radian)],
                              [np.sin(radian), np.cos(radian)]])

    # Melakukan rotasi pada pusat lingkaran
    center_rotated = np.dot(center, rotasi_matrix)
    return center_rotated

# Pusat lingkaran dan radius (langsung diinput)
center = np.array([3, 2])
radius = 2

# Sudut rotasi (dalam derajat)
sudut_rotasi = 60

# --- Plot 1: Lingkaran Sebelum Rotasi ---
# Menggambar lingkaran sebelum rotasi
fig, ax = plt.subplots(figsize=(8, 8))
ax.set_xlim(-10, 10)
ax.set_ylim(-10, 10)
ax.axhline(0, color='black', linewidth=0.5, ls='--')
ax.axvline(0, color='black', linewidth=0.5, ls='--')
ax.grid(color='gray', linestyle='--', linewidth=0.5)

# Gambar lingkaran asli
gambar_lingkaran(radius, center, ax, 'Lingkaran Asli')
ax.plot(center[0], center[1], 'ro', markersize=8, label='Pusat')  # Tandai pusat
ax.set_title('Lingkaran Sebelum Rotasi')
ax.set_aspect('equal', adjustable='box')
ax.legend()
plt.show()

# --- Proses Rotasi ---
# Melakukan rotasi pada pusat lingkaran
center_rotated = rotasi_lingkaran(center, sudut_rotasi)

# --- Plot 2: Lingkaran Setelah Rotasi ---
# Menggambar lingkaran setelah rotasi
fig, ax = plt.subplots(figsize=(8, 8))
ax.set_xlim(-10, 10)
ax.set_ylim(-10, 10)
ax.axhline(0, color='black', linewidth=0.5, ls='--')
ax.axvline(0, color='black', linewidth=0.5, ls='--')
ax.grid(color='gray', linestyle='--', linewidth=0.5)

# Gambar lingkaran yang sudah dirotasi
gambar_lingkaran(radius, center_rotated, ax, 'Lingkaran Rotasi')
ax.plot(center_rotated[0], center_rotated[1], 'go', markersize=8, label='Pusat Rotasi')  # Tandai pusat baru
ax.set_title(f'Lingkaran Setelah Rotasi ({sudut_rotasi} Derajat)')
ax.set_aspect('equal', adjustable='box')
ax.legend()
plt.show()
```

```python
# Import library yang diperlukan
import numpy as np
import matplotlib.pyplot as plt

# Fungsi untuk menggambar segi enam
def gambar_segi_enam(titik, ax):
    # Menutup bentuk segi enam dengan mengulangi titik pertama di akhir
    # axis=0 artinya menambahkan baris baru ke array
    titik_plot = np.append(titik, [titik[0]], axis=0)
    ax.plot(titik_plot[:, 0], titik_plot[:, 1], label='Segi Enam', linewidth=2)

# Fungsi untuk melakukan rotasi
def rotasi(titik, sudut):
    # Menghitung sudut dalam radian
    radian = np.radians(sudut)
    
    # Matriks rotasi
    rotasi_matrix = np.array([[np.cos(radian), -np.sin(radian)],
                              [np.sin(radian), np.cos(radian)]])
    
    # Menghitung titik yang sudah dirotasi menggunakan perkalian dot matrix
    titik_rotated = np.dot(titik, rotasi_matrix)
    return titik_rotated

# Titik-titik segi enam (langsung diinput)
# Membuat segi enam beraturan dengan pusat di (3, 2) dan radius 2
center = np.array([3, 2])
radius = 2
angles = np.linspace(0, 2*np.pi, 7)[:-1]  # 6 titik, tidak termasuk titik terakhir
titik_asli = np.array([[center[0] + radius * np.cos(angle), 
                        center[1] + radius * np.sin(angle)] for angle in angles])

# Sudut rotasi (dalam derajat)
sudut_rotasi = 45

# --- Plot 1: Segi Enam Sebelum Rotasi ---
fig, ax = plt.subplots(figsize=(8, 8))
ax.set_xlim(-10, 10)
ax.set_ylim(-10, 10)
ax.axhline(0, color='black', linewidth=0.5, ls='--')
ax.axvline(0, color='black', linewidth=0.5, ls='--')
ax.grid(color='gray', linestyle='--', linewidth=0.5)

# Gambar segi enam asli
gambar_segi_enam(titik_asli, ax)
ax.set_title('Segi Enam Sebelum Rotasi')
ax.set_aspect('equal', adjustable='box')
ax.legend()
plt.show()

# --- Proses Rotasi ---
# Melakukan rotasi
titik_rotated = rotasi(titik_asli, sudut_rotasi)

# --- Plot 2: Segi Enam Setelah Rotasi ---
fig, ax = plt.subplots(figsize=(8, 8))
ax.set_xlim(-10, 10)
ax.set_ylim(-10, 10)
ax.axhline(0, color='black', linewidth=0.5, ls='--')
ax.axvline(0, color='black', linewidth=0.5, ls='--')
ax.grid(color='gray', linestyle='--', linewidth=0.5)

# Gambar segi enam yang sudah dirotasi
gambar_segi_enam(titik_rotated, ax)
ax.set_title(f'Segi Enam Setelah Rotasi ({sudut_rotasi} Derajat)')
ax.set_aspect('equal', adjustable='box')
ax.legend()
plt.show()
```
