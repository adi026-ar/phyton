```python
import matplotlib.pyplot as plt
import numpy as np

# Fungsi untuk menggambar persegi panjang sebelum dan sesudah shear
def plot_persegi_panjang(original_points, sheared_points):
    plt.figure(figsize=(10, 5))
    
    # Gambar Persegi Panjang Asli
    plt.subplot(1, 2, 1)
    # Menambahkan titik awal di akhir agar garis menyambung (closed loop)
    plt.plot(*zip(*original_points, original_points[0]), marker='o', color='blue')
    plt.fill(*zip(*original_points), alpha=0.3, color='blue') # Opsional: memberi warna isi
    plt.title('Persegi Panjang Asli')
    plt.grid(True)
    plt.axhline(0, color='black', linewidth=1)
    plt.axvline(0, color='black', linewidth=1)
    
    # Gambar Persegi Panjang Setelah Shear
    plt.subplot(1, 2, 2)
    plt.plot(*zip(*sheared_points, sheared_points[0]), marker='o', color='red')
    plt.fill(*zip(*sheared_points), alpha=0.3, color='red') # Opsional: memberi warna isi
    plt.title('Persegi Panjang Setelah Shear')
    plt.grid(True)
    plt.axhline(0, color='black', linewidth=1)
    plt.axvline(0, color='black', linewidth=1)
    
    plt.show()

# Fungsi untuk melakukan shear pada objek persegi panjang
def shear_persegi_panjang(points, shear_x, shear_y):
    # Matriks transformasi shear
    shear_matrix = np.array([[1, shear_x],
                             [shear_y, 1]])
    
    sheared_points = []
    # Mengalikan setiap titik dengan matriks shear
    for point in points:
        new_point = np.dot(shear_matrix, point)
        sheared_points.append(new_point)
        
    return np.array(sheared_points)

# --- PROGRAM UTAMA ---

# 1. Nilai shear otomatis (bisa diubah sesuai kebutuhan)
print("=== Tugas Transformasi Shear Persegi Panjang ===")
shear_x = 0.5  # Faktor shear horizontal
shear_y = 0.5  # Faktor shear vertikal
print(f"Shear X: {shear_x}")
print(f"Shear Y: {shear_y}")

# 2. Koordinat asli persegi panjang (berdasarkan gambar tugas)
# x dari 0 sampai 8, y dari 0 sampai 4
persegi_panjang_points = np.array([
    [0, 0], # Titik A
    [8, 0], # Titik B
    [8, 4], # Titik C
    [0, 4]  # Titik D
])

# 3. Lakukan transformasi shear
sheared_points = shear_persegi_panjang(persegi_panjang_points, shear_x, shear_y)

# 4. Tampilkan hasil
plot_persegi_panjang(persegi_panjang_points, sheared_points)
```
