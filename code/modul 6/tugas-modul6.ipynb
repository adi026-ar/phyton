```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

def draw_cylinder(ax, center, radius, height, title):
    """Fungsi untuk menggambar tabung (cylinder) 3D"""
    # Membuat data untuk permukaan silinder
    theta = np.linspace(0, 2 * np.pi, 50)
    z = np.linspace(0, height, 50)
    theta_grid, z_grid = np.meshgrid(theta, z)
    
    # Koordinat silinder
    x = center[0] + radius * np.cos(theta_grid)
    y = center[1] + radius * np.sin(theta_grid)
    
    # Menggambar permukaan silinder
    ax.plot_surface(x, y, z_grid, alpha=0.7, color='gray')
    
    # Menggambar tutup atas dan bawah
    theta_cap = np.linspace(0, 2 * np.pi, 50)
    r_cap = np.linspace(0, radius, 20)
    theta_cap_grid, r_cap_grid = np.meshgrid(theta_cap, r_cap)
    
    x_cap = center[0] + r_cap_grid * np.cos(theta_cap_grid)
    y_cap = center[1] + r_cap_grid * np.sin(theta_cap_grid)
    
    # Tutup bawah (z=0)
    z_bottom = np.zeros_like(x_cap)
    ax.plot_surface(x_cap, y_cap, z_bottom, alpha=0.7, color='gray')
    
    # Tutup atas (z=height)
    z_top = np.ones_like(x_cap) * height
    ax.plot_surface(x_cap, y_cap, z_top, alpha=0.7, color='gray')
    
    # Pengaturan plot
    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    ax.set_zlabel('Z')
    ax.set_title(title)
    ax.set_xlim([-10, 10])
    ax.set_ylim([-10, 10])
    ax.set_zlim([0, 10])

def scale_cylinder(radius, height, scale_factor):
    """Fungsi untuk melakukan penskalaan tabung"""
    new_radius = radius * scale_factor
    new_height = height * scale_factor
    return new_radius, new_height

# Data tabung
center = (0, 0)  # Pusat tabung di (x, y)
radius = 2       # Jari-jari tabung
height = 5       # Tinggi tabung
scale_factor = 2.0  # Faktor skala

# Membuat figure dengan 2 subplot
fig = plt.figure(figsize=(14, 6))

# Subplot 1: Tabung Asli
ax1 = fig.add_subplot(121, projection='3d')
draw_cylinder(ax1, center, radius, height, "Tabung Asli")

# Melakukan penskalaan
new_radius, new_height = scale_cylinder(radius, height, scale_factor)

# Subplot 2: Tabung Setelah Penskalaan
ax2 = fig.add_subplot(122, projection='3d')
draw_cylinder(ax2, center, new_radius, new_height, f"Tabung Setelah Penskalaan (Faktor Skala: {scale_factor})")

plt.tight_layout()
plt.show()
```
