```python
import matplotlib.pyplot as plt

def draw_triangle(vertices, title):
    """Fungsi untuk menggambar segitiga berdasarkan titik-titik yang diberikan."""
    # Mengurai titik-titik menjadi koordinat x dan y
    x, y = zip(*vertices)
    
    # Konversi tuple ke list untuk memungkinkan concatenation
    x = list(x)
    y = list(y)

    # Menutup segitiga dengan menambahkan titik pertama di akhir
    x.append(x[0])
    y.append(y[0])

    plt.figure()
    plt.plot(x, y, marker='o') # Menggambar segitiga dengan marker di setiap titik
    plt.title(title)
    plt.xlim(-100, 100) # Mengatur batas sumbu x
    plt.ylim(-100, 100) # Mengatur batas sumbu y
    plt.axhline(0, color='black', linewidth=0.5, ls='--') # Garis horizontal di sumbu x
    plt.axvline(0, color='black', linewidth=0.5, ls='--') # Garis vertikal di sumbu y
    plt.grid() # Menampilkan grid
    plt.gca().set_aspect('equal', adjustable='box') # Memastikan sumbu x dan y memiliki skala yang sama
    plt.show()

def scale_triangle(vertices, scale_factor):
    """Fungsi untuk melakukan penskalaan segitiga berdasarkan faktor skala yang diberikan."""
    # Menghitung titik-titik segitiga yang sudah dipenskala
    scaled_vertices = [(x * scale_factor, y * scale_factor) for x, y in vertices]
    return scaled_vertices

# Nilai input langsung
x1, y1 = 10, 20
x2, y2 = 30, 15
x3, y3 = 25, 40
scale_factor = 2

# Mendefinisikan titik-titik segitiga
triangle_vertices = [(x1, y1), (x2, y2), (x3, y3)]

# Menggambar segitiga asli
draw_triangle(triangle_vertices, "Segitiga Asli")

# Melakukan penskalaan pada segitiga
scaled_triangle_vertices = scale_triangle(triangle_vertices, scale_factor)

# Menggambar segitiga setelah penskalaan
draw_triangle(scaled_triangle_vertices, f"Segitiga Setelah Penskalaan (Faktor Skala: {scale_factor})")
```

```python
import numpy as np
import matplotlib.pyplot as plt

def draw_circle(center, radius, title):
    """Fungsi untuk menggambar lingkaran berdasarkan pusat dan radius."""
    theta = np.linspace(0, 2 * np.pi, 100) # Membuat sudut dari 0 hingga 2pi
    x = center[0] + radius * np.cos(theta) # Menghitung koordinat x
    y = center[1] + radius * np.sin(theta) # Menghitung koordinat y

    plt.figure()
    plt.plot(x, y) # Menggambar lingkaran
    plt.title(title)
    plt.xlim(-10, 10) # Mengatur batas sumbu x
    plt.ylim(-10, 10) # Mengatur batas sumbu y
    plt.axhline(0, color='black', linewidth=0.5, ls='--') # Garis horizontal di sumbu x
    plt.axvline(0, color='black', linewidth=0.5, ls='--') # Garis vertikal di sumbu y
    plt.grid() # Menampilkan grid
    plt.gca().set_aspect('equal', adjustable='box') # Memastikan sumbu x dan y memiliki skala yang sama
    plt.show()

def scale_circle(center, radius, scale_factor):
    """Fungsi untuk melakukan penskalaan lingkaran berdasarkan faktor skala yang diberikan."""
    new_radius = radius * scale_factor # Menghitung radius baru
    return new_radius

# Nilai input langsung
center_x, center_y = 0, 0
radius = 3
scale_factor = 2

# Mendefinisikan pusat lingkaran
circle_center = (center_x, center_y)

# Menggambar lingkaran asli
draw_circle(circle_center, radius, "Lingkaran Asli")

# Melakukan penskalaan pada lingkaran
new_radius = scale_circle(circle_center, radius, scale_factor)

# Menggambar lingkaran setelah penskalaan
draw_circle(circle_center, new_radius, f"Lingkaran Setelah Penskalaan (Faktor Skala: {scale_factor})")
```

```python
import numpy as np
import matplotlib.pyplot as plt

def draw_circle(center, radius, title):
    """Fungsi untuk menggambar lingkaran berdasarkan pusat dan radius."""
    theta = np.linspace(0, 2 * np.pi, 100) # Membuat sudut dari 0 hingga 2pi
    x = center[0] + radius * np.cos(theta) # Menghitung koordinat x
    y = center[1] + radius * np.sin(theta) # Menghitung koordinat y

    plt.plot(x, y)
    plt.title(title)
    plt.xlim(-10, 10) # Mengatur batas sumbu x
    plt.ylim(-10, 10) # Mengatur batas sumbu y
    plt.axhline(0, color='black', linewidth=0.5, ls='--') # Garis horizontal
    plt.axvline(0, color='black', linewidth=0.5, ls='--') # Garis vertikal
    plt.grid()
    plt.gca().set_aspect('equal', adjustable='box')

def draw_rectangle(bottom_left, width, height, title):
    """Fungsi untuk menggambar persegi panjang berdasarkan titik kiri bawah, lebar, dan tinggi."""
    x = [bottom_left[0], bottom_left[0] + width, bottom_left[0] + width, bottom_left[0], bottom_left[0]]
    y = [bottom_left[1], bottom_left[1], bottom_left[1] + height, bottom_left[1] + height, bottom_left[1]]

    plt.plot(x, y)
    plt.title(title)
    plt.xlim(-10, 10)
    plt.ylim(-10, 10)
    plt.axhline(0, color='black', linewidth=0.5, ls='--')
    plt.axvline(0, color='black', linewidth=0.5, ls='--')
    plt.grid()
    plt.gca().set_aspect('equal', adjustable='box')

def scale_circle(center, radius, scale_factor):
    """Fungsi penskalaan lingkaran."""
    new_radius = radius * scale_factor
    return new_radius

def scale_rectangle(bottom_left, width, height, scale_factor):
    """Fungsi penskalaan persegi panjang (mengubah lebar dan tinggi)."""
    new_width = width * scale_factor
    new_height = height * scale_factor
    return new_width, new_height

# --- INPUT DATA LANGSUNG ---
# Data Lingkaran
circle_center_x, circle_center_y = 2, 3
circle_radius = 4
scale_factor_circle = 1.5

# Data Persegi Panjang
rect_bottom_left_x, rect_bottom_left_y = -3, -2
rect_width = 4
rect_height = 3
scale_factor_rectangle = 2

# Definisi Variabel
circle_center = (circle_center_x, circle_center_y)
rect_bottom_left = (rect_bottom_left_x, rect_bottom_left_y)

# --- PLOTTING ---
plt.figure(figsize=(12, 6)) # Membuat ukuran gambar lebih lebar

# Subplot 1: Lingkaran
plt.subplot(1, 2, 1) # 1 baris, 2 kolom, gambar ke-1
draw_circle(circle_center, circle_radius, "Lingkaran Asli")
new_radius = scale_circle(circle_center, circle_radius, scale_factor_circle)
draw_circle(circle_center, new_radius, f"Lingkaran Skala {scale_factor_circle}")

# Subplot 2: Persegi Panjang
plt.subplot(1, 2, 2) # 1 baris, 2 kolom, gambar ke-2
draw_rectangle(rect_bottom_left, rect_width, rect_height, "Persegi Panjang Asli")
new_width, new_height = scale_rectangle(rect_bottom_left, rect_width, rect_height, scale_factor_rectangle)
draw_rectangle(rect_bottom_left, new_width, new_height, f"Persegi Panjang Skala {scale_factor_rectangle}")

plt.tight_layout()
plt.show()
```
