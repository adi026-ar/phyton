```python
import matplotlib.pyplot as plt

# Fungsi untuk menggambar segitiga dengan titik-titik sudutnya
def gambar_segitiga(titik, color='blue', label='Segitiga'):
    # Menyusun titik segitiga (menambahkan kembali titik awal agar garis tertutup)
    x = [titik[0][0], titik[1][0], titik[2][0], titik[0][0]]
    y = [titik[0][1], titik[1][1], titik[2][1], titik[0][1]]
    
    # Menggambar segitiga
    plt.plot(x, y, marker='o', color=color, linewidth=2, markersize=8, label=label)
    plt.fill(x, y, alpha=0.3, color=color) # Mengisi segitiga dengan warna transparan
    
    # Menambahkan label koordinat pada setiap titik
    for i, (px, py) in enumerate(titik):
        plt.text(px, py, f'  ({px:.1f}, {py:.1f})', fontsize=9, ha='left', va='bottom')

# Fungsi untuk melakukan pencerminan
def pencerminan(titik, sumbu):
    # Melakukan pencerminan berdasarkan sumbu
    if sumbu == 'x':
        # Pencerminan terhadap sumbu x
        return [(x, -y) for (x, y) in titik]
    elif sumbu == 'y':
        # Pencerminan terhadap sumbu y
        return [(-x, y) for (x, y) in titik]
    elif sumbu == 'origin':
        # Pencerminan terhadap origin
        return [(-x, -y) for (x, y) in titik]
    else:
        raise ValueError("Sumbu tidak valid. Pilih 'x', 'y', atau 'origin'.")

# Titik segitiga (otomatis)
titik_segitiga = [(2, 3), (5, 3), (3.5, 6)]

# Jenis sumbu pencerminan (otomatis)
sumbu = 'x'  # Bisa diganti dengan 'y' atau 'origin'

print(f"Titik segitiga: {titik_segitiga}")
print(f"Sumbu pencerminan: {sumbu}")

# Menggambar segitiga asli
plt.figure(figsize=(10, 10))
plt.title("Pencerminan Segitiga 2D")
plt.axhline(0, color='black', linewidth=0.5, ls='--')
plt.axvline(0, color='black', linewidth=0.5, ls='--')
plt.grid(color='gray', linestyle='--', linewidth=0.5)

# Gambar segitiga asli
gambar_segitiga(titik_segitiga, color='blue', label='Segitiga Asli')

# Pencerminan segitiga
titik_segitiga_pencerminan = pencerminan(titik_segitiga, sumbu)

# Menggambar segitiga setelah pencerminan
gambar_segitiga(titik_segitiga_pencerminan, color='red', label='Segitiga Pencerminan')

# Menambahkan legend dan menampilkan plot
plt.legend()
plt.xlim(-10, 10)
plt.ylim(-10, 10)
plt.gca().set_aspect('equal', adjustable='box')
plt.show()
```

```python
import matplotlib.pyplot as plt

# Fungsi untuk menggambar persegi panjang dengan titik-titik sudutnya
def gambar_persegi_long_panjang(ax, x, y, lebar, tinggi, label='Persegi Panjang', color='blue'):
    """Fungsi untuk menggambar persegi panjang pada grafik."""
    persegi_panjang = plt.Rectangle((x, y), lebar, tinggi, fill=None, edgecolor=color, linewidth=2, label=label)
    ax.add_patch(persegi_panjang)
    
    # Menggambar titik-titik sudut
    titik_x = [x, x + lebar, x + lebar, x, x]
    titik_y = [y, y, y + tinggi, y + tinggi, y]
    ax.plot(titik_x, titik_y, 'o', color=color, markersize=8)
    
    # Menambahkan label koordinat pada setiap titik
    ax.text(x, y, f'({x:.1f}, {y:.1f})', fontsize=8, ha='right', va='top')
    ax.text(x + lebar, y, f'({x + lebar:.1f}, {y:.1f})', fontsize=8, ha='left', va='top')
    ax.text(x + lebar, y + tinggi, f'({x + lebar:.1f}, {y + tinggi:.1f})', fontsize=8, ha='left', va='bottom')
    ax.text(x, y + tinggi, f'({x:.1f}, {y + tinggi:.1f})', fontsize=8, ha='right', va='bottom')

# Fungsi untuk mencerminkan persegi panjang
def pencerminan_persegi_panjang(x, y, lebar, tinggi, sumbu):
    """Fungsi untuk melakukan pencerminan persegi panjang."""
    if sumbu == 'X':
        # Pencerminan terhadap sumbu X
        return x, -y - tinggi, lebar, tinggi
    elif sumbu == 'Y':
        # Pencerminan terhadap sumbu Y
        return -x - lebar, y, lebar, tinggi
    elif sumbu == 'Origin':
        # Pencerminan terhadap origin
        return -x - lebar, -y - tinggi, lebar, tinggi

# Input otomatis
x = 2.0
y = 1.0
lebar = 4.0
tinggi = 3.0
sumbu = 'X'  # Bisa diganti dengan 'Y' atau 'Origin'

print(f"Koordinat X sudut kiri bawah: {x}")
print(f"Koordinat Y sudut kiri bawah: {y}")
print(f"Lebar persegi panjang: {lebar}")
print(f"Tinggi persegi panjang: {tinggi}")
print(f"Sumbu pencerminan: {sumbu}")

# Membuat plot
fig, ax = plt.subplots(figsize=(10, 10))

# Menggambar persegi panjang asli
gambar_persegi_long_panjang(ax, x, y, lebar, tinggi, label='Persegi Panjang Asli', color='blue')

# Menggambar persegi panjang setelah pencerminan
x_baru, y_baru, lebar_baru, tinggi_baru = pencerminan_persegi_panjang(x, y, lebar, tinggi, sumbu)
gambar_persegi_long_panjang(ax, x_baru, y_baru, lebar_baru, tinggi_baru, label='Persegi Panjang Setelah Pencerminan', color='red')

# Menambahkan detail ke plot
ax.set_xlim(-10, 10)
ax.set_ylim(-10, 10)
ax.set_aspect('equal', adjustable='box')
ax.axhline(0, color='black', linewidth=0.5, ls='--') # Sumbu X
ax.axvline(0, color='black', linewidth=0.5, ls='--') # Sumbu Y
plt.grid(color='gray', linestyle='--', linewidth=0.5)
plt.legend()
plt.title('Pencerminan Persegi Panjang 2D')
plt.show()
```
