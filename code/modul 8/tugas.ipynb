```python
import matplotlib.pyplot as plt

# Fungsi untuk menggambar belah ketupat dengan label titik
def gambar_belah_ketupat(titik, label, warna):
    # Menambahkan titik awal di akhir agar poligon tertutup
    x = [p[0] for p in titik] + [titik[0][0]]
    y = [p[1] for p in titik] + [titik[0][1]]
    plt.plot(x, y, color=warna, marker='o', markersize=8, linewidth=2)
    plt.fill(x, y, alpha=0.3, color=warna, label=label)
    
    # Menambahkan label koordinat pada setiap titik
    for i, (px, py) in enumerate(titik):
        plt.text(px, py, f'  ({px:.1f}, {py:.1f})', fontsize=9, ha='left', va='bottom')

# Fungsi pencerminan terhadap sumbu X
def pencerminan_sumbu_x(titik):
    # Rumus pencerminan sumbu X: (x, y) -> (x, -y)
    return [(x, -y) for (x, y) in titik]

# Mendefinisikan titik-titik belah ketupat (di kuadran atas)
# Titik: atas, kanan, bawah, kiri
titik_asli = [(0, 5), (2, 3), (0, 1), (-2, 3)]

# Melakukan pencerminan
titik_tercermin = pencerminan_sumbu_x(titik_asli)

# Membuat plot
fig, ax = plt.subplots(figsize=(10, 10))

# Menggambar objek asli dan hasil pencerminan
gambar_belah_ketupat(titik_asli, 'Belah Ketupat Asli', 'blue')
gambar_belah_ketupat(titik_tercermin, 'Belah Ketupat Setelah Pencerminan', 'red')

# Pengaturan detail grafik (sesuai template di gambar)
ax.set_xlim(-10, 10)
ax.set_ylim(-10, 10)
ax.set_aspect('equal', adjustable='box')
ax.axhline(0, color='black', linewidth=0.5, ls='--') # Sumbu X
ax.axvline(0, color='black', linewidth=0.5, ls='--') # Sumbu Y
plt.grid(color='gray', linestyle='--', linewidth=0.5)
plt.legend()
plt.title('Pencerminan Belah Ketupat terhadap Sumbu X')
plt.show()
```
